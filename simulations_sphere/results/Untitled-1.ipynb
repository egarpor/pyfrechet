```python
import numpy as np
from scipy.special import iv  # Modified Bessel function of the first kind
from scipy.integrate import nquad  # Numerical integration

# Von Mises-Fisher density for d=3
def von_mises_fisher_density(v, mu, kappa):
    """
    Compute the von Mises-Fisher density for d=3.
    
    Parameters:
    v (array-like): A point on the unit sphere S^2 (3D vector).
    mu (array-like): Mean direction (3D unit vector).
    kappa (float): Concentration parameter (kappa >= 0).
    
    Returns:
    float: The density value at v.
    """
    # Ensure v and mu are unit vectors
    v = np.array(v) / np.linalg.norm(v)
    mu = np.array(mu) / np.linalg.norm(mu)
    
    # Normalizing constant for d=3
    C3 = kappa / (4 * np.pi * np.sinh(kappa))
    
    # Density formula
    density = C3 * np.exp(kappa * np.dot(mu, v))
    
    return density

# Function to integrate over the unit sphere S^2
def integrand(theta, phi, mu, kappa):
    """
    Integrand for the von Mises-Fisher density in spherical coordinates.
    
    Parameters:
    theta (float): Polar angle (0 to pi).
    phi (float): Azimuthal angle (0 to 2pi).
    mu (array-like): Mean direction (3D unit vector).
    kappa (float): Concentration parameter (kappa >= 0).
    
    Returns:
    float: The density value at (theta, phi) multiplied by the Jacobian (sin(theta)).
    """
    # Convert spherical coordinates to Cartesian coordinates
    x = np.sin(theta) * np.cos(phi)
    y = np.sin(theta) * np.sin(phi)
    z = np.cos(theta)
    v = np.array([x, y, z])
    
    # Compute the density and multiply by the Jacobian (sin(theta))
    return von_mises_fisher_density(v, mu, kappa) * np.sin(theta)

# Parameters
mu = np.array([1, 0, 0])  # Mean direction (unit vector)
kappa = 100.0  # Concentration parameter

# Numerical integration over the unit sphere S^2
# theta ranges from 0 to pi, phi ranges from 0 to 2*pi
integral, _ = nquad(integrand, [(0, np.pi), (0, 2 * np.pi)], args=(mu, kappa))

# Check if the integral is approximately 1
print(f"Integral of the von Mises-Fisher density over S^2: {integral}")
if np.isclose(integral, 1.0, atol=1e-4):
    print("The density integrates to 1 (within numerical tolerance).")
else:
    print("The density does not integrate to 1.")
```

```python
import numpy as np
import scipy.integrate as integrate
import scipy.special as sp

def hyperbolic_inner_product(x, y):
    """Computes the hyperbolic inner product x * y."""
    return x[0] * y[0] - np.sum(x[1:] * y[1:])

def hyperboloid_density(u, d, kappa, psi):
    """Computes the density function of the von Mises-Fisher-like distribution on H_d."""
    A_d = (kappa**(d/2 - 1)) / ((2 * np.pi) ** (d/2 - 1) * 0.12533)
    v = np.array([np.cosh(u)] + [np.sinh(u)] * (d - 1))  # A point on H_d
    return A_d * np.exp(-kappa * hyperbolic_inner_product(psi, v)) * (np.sinh(u) ** (d - 2))

def integrate_density(d, kappa, psi):
    """Numerically integrates the density over the unit hyperboloid H_d to check normalization."""
    result, error = integrate.quad(lambda u: hyperboloid_density(u, d, kappa, psi), 0.01, 100)
    return result, error

# Example: Checking normalization for d = 3, kappa = 1, and psi = (1,0,0)
d = 3
kappa = 10
psi = np.array([1, 0, 0])  # Example mean direction on H_d
integral, error = integrate_density(d, kappa, psi)
print(f"Integral result: {integral}, Estimated error: {error}")
```
